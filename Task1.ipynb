```python
#Missing values
df.isnull().sum()
```

```python
#Handling missing values
df['Income'].fillna(df['Income'].median(), inplace=True)
```

```python
#Check duplicate records
df.duplicated().sum()
```

```python
df.dtypes
```

```python
#Convert date column
df['Dt_Customer']=pd.to_datetime(df['Dt_Customer'], dayfirst=True)
```

```python
df['Dt_Customer'].head()
```

```python
df.info()
```

```python
df['Age'] = 2026 - df['Year_Birth']
```

```python
df.columns
```

```python
df['Total_Spending']=(
    df['MntWines']
    + df['MntFruits']
    + df['MntMeatProducts']
    + df['MntFishProducts']
    + df['MntSweetProducts']
    + df['MntGoldProds']
)
```

```python
df.to_csv('marketing_campaign_cleaned.csv', index=False)
```
