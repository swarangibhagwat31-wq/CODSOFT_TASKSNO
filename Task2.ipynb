```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv("marketing_campaign_cleaned.csv")
```

```python
# Age distribution 
df['Age'].describe()
```

```python
df['Age'].head()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=20)
plt.title('Age Distribution of Customers')
plt.show()
```

Insight:
-Most customers are between 45 and 65 years old, with the highest concentration around 50–60 years.
-Very few customers are above 75 years. 
-An outlier above 120 years is also observed, which may represent an incorrect record and should be reviewed.

```python
# Income Distribution
plt.figure(figsize=(8,5))
sns.histplot(df['Income'], bins=30)
plt.title('Income Distribution')
plt.show()
```

Insight:
-The majority of customers belong to the middle-income segment, with most incomes ranging between 30,000 and 90,000.
-The income distribution is right-skewed, indicating that a small number of customers have exceptionally high incomes.
-These high-income customers may represent a valuable target segment for premium marketing campaigns.

```python
# Education Analysis
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Education')
plt.xticks(rotation=45)
plt.show()
```

Insight:

```python
# Product Spending analysis
spending={
    'Wine': df['MntWines'].sum(),
    'Fruits': df['MntFruits'].sum(),
    'Meat': df['MntMeatProducts'].sum(),
    'Fish': df['MntFishProducts'].sum(),
    'Sweets': df['MntSweetProducts'].sum(),
    'Gold': df['MntGoldProds'].sum()
}
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.bar(spending.keys(), spending.values())
plt.title('Total Spending by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Spending')
plt.show()
```

Insights:
-Wine is the most preferred product category and generates the highest revenue.
-Meat products are the second-largest contributor to total spending.
-Fruits and Sweets contribute the least to customer spending.
-Customer purchasing behavior is heavily concentrated on Wine and Meat products.
-Promotional strategies can target low-spending categories to improve sales.

```python
# Purchase channel analysis
purchase_channels= df[['NumWebPurchases','NumCatalogPurchases','NumStorePurchases']].sum()
```

```python
purchase_channels.plot(kind='bar', figsize=(8,5))
plt.title("Purchase Channel Analysis")
plt.xlabel("Purchase Channel")
plt.ylabel("Total Purchases")
plt.show()
```

Insights:
-Store purchases are the highest among all purchase channels, indicating that customers prefer shopping from physical stores.
-Web purchases are the second most popular channel, showing a significant level of online engagement.
-Catalog purchases have the lowest number of purchases compared to stores and websites.
-Customers appear to favor direct purchasing methods over catalog-based shopping.


```python
# Correlation analysis
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(numeric_only=True), cmap='coolwarm')
plt.show()
```

Insights:
-Income has a positive relationship with Total_Spending, indicating that higher-income customers tend to spend more.
-Total_Spending is positively correlated with purchases across different product categories, especially Wine and Meat products.
-Store, Web, and Catalog purchases show positive correlations with Total_Spending, suggesting that frequent buyers spend more overall.
-NumWebVisitsMonth shows a weak negative relationship with purchases, indicating that more website visits do not always result in more purchases.
-Age has a weak correlation with Total_Spending, suggesting that spending behavior is not strongly influenced by age.
-Product spending categories (Wine, Fruits, Meat, Fish, Sweets, and Gold) are positively correlated with one another, indicating that customers who spend more in one category often spend more in others.
